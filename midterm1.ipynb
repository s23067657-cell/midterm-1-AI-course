```python
import math, pandas as pd

ranks = pd.read_csv("my_data/retrieval_ranks.csv")
print(ranks.columns.tolist())
lang_col = [c for c in ranks.columns
            if ranks[c].astype(str).isin(["EN","RU","KK"]).any()][0]

def wilson(h, n, z=1.96):
    p = h/n; d = 1 + z*z/n
    c = (p + z*z/(2*n))/d
    w = z*math.sqrt(p*(1-p)/n + z*z/(4*n*n))/d
    return round(c-w,3), round(c+w,3)

def mrr(s): return sum(1/r for r in s if r <= 10)/len(s)

print("\n=== 1a ===")
for cfg in "AB":
    s = ranks[f"rank_{cfg}"]
    print(cfg, "recall@1/3/5:", [round((s<=k).mean(),4) for k in (1,3,5)],
          "MRR@10:", round(mrr(s),4))

print("\n=== 1b ===")
for cfg in "AB":
    for k in (3,5):
        s = ranks[f"rank_{cfg}"]; h = int((s<=k).sum())
        print(f"({cfg},{k}) hits={h}/60 recall={h/60:.4f} Wilson={wilson(h,60)}")

share = {"EN":.34,"RU":.40,"KK":.26}
q  = {"EN":31,"RU":44,"KK":105}
a  = {"EN":139,"RU":196,"KK":470}
ch = {"A":{"EN":269,"RU":379,"KK":909}, "B":{"EN":80,"RU":113,"KK":270}}
SYS, PIN, POUT, REQ, BUDGET = 197, 3, 15, 7000, 19700

def annual(cfg, k):
    tot = 0
    for l in share:
        inp = SYS + q[l] + k*ch[cfg][l]
        tot += share[l]*(inp*PIN + a[l]*POUT)/1e6
    return tot*REQ*365

print("\n=== 1c (budget $19,700) ===")
for cfg in "AB":
    for k in (3,5):
        c = annual(cfg,k)
        print(f"({cfg},{k}) ${c:,.0f}", "OK" if c <= BUDGET else "OVER BUDGET")

print("\n=== per-language recall (n=20) ===")
for cfg in "AB":
    for k in (3,5):
        for l in ["EN","RU","KK"]:
            s = ranks[ranks[lang_col]==l][f"rank_{cfg}"]; h = int((s<=k).sum())
            print(f"({cfg},{k}) {l}: {h}/20 = {h/20:.2f}, Wilson={wilson(h,20)}")
```
